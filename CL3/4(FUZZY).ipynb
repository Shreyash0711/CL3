```python
# Define fuzzy sets
A = {'x1': 0.2, 'x2': 0.7, 'x3': 0.5}
B = {'x1': 0.6, 'x2': 0.4, 'x3': 0.8}
# Union
union = {k: max(A[k], B[k]) for k in A}
# Intersection
intersection = {k: min(A[k], B[k]) for k in A}
# Complement of A
complement_A = {k: 1 - A[k] for k in A}
# Difference (A - B)
difference = {k: min(A[k], 1 - B[k]) for k in A}
print("Union:", union)
print("Intersection:", intersection)
print("Complement of A:", complement_A)
print("Difference (A-B):", difference)
```

```python
# Cartesian product (Relation R)
R = {}
for a in A:
    for b in B:
        R[(a, b)] = min(A[a], B[b])
print("\nFuzzy Relation R:")
for key, val in R.items():
    print(f"{key}: {val}")
```

```python
C = {'y1': 0.3, 'y2': 0.9, 'y3': 0.6}
# Relation S
S = {}
for b in B:
    for c in C:
        S[(b, c)] = min(B[b], C[c])
```

```python
# Max-Min Composition
T = {}
for a in A:
    for c in C:
        values = []
        for b in B:
            values.append(min(R[(a, b)], S[(b, c)]))
        T[(a, c)] = max(values)
print("\nMax-Min Composition T:")
for key, val in T.items():
    print(f"{key}: {val}")
```
